```python

# Calculate store-wise IQR boundaries

def find_outliers(group):

    q1 = group['Weekly_Sales'].quantile(0.25)

    q3 = group['Weekly_Sales'].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr

    upper = q3 + 1.5 * iqr

    return group[
        (group['Weekly_Sales'] < lower) |
        (group['Weekly_Sales'] > upper)
    ]

# Apply IQR separately to every store

store_outliers = (
    df.groupby('Store', group_keys=False)
    .apply(find_outliers)
)

print("Store-wise Outliers:", len(store_outliers))

display(store_outliers.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
df['Date'] = pd.to_datetime(
    df['Date'],
    format='%d-%m-%Y'
)
```

```python
df = df.sort_values(['Store', 'Date'])

df.head()
```

```python
store_counts = df.groupby('Store').size()

print(store_counts)
print("Total stores:", df['Store'].nunique())
```

```python
store_20 = (
    df[df['Store'] == 20]
    .sort_values('Date')
    .set_index('Date')['Weekly_Sales']
)

print(store_20.head())
print(store_20.tail())
```

```python
plt.figure(figsize=(14, 5))

plt.plot(store_20)

plt.title('Weekly Sales Trend - Store 20')
plt.xlabel('Date')
plt.ylabel('Weekly Sales')

plt.grid(True)
plt.show()
```

```python
train = store_20.iloc[:-12]
test = store_20.iloc[-12:]

print("Training records:", len(train))
print("Testing records:", len(test))

print("Training ends:", train.index.max())
print("Testing starts:", test.index.min())
```

```python
last_sales = train.iloc[-1]

naive_predictions = pd.Series(
    [last_sales] * len(test),
    index=test.index
)

print(naive_predictions)
```

```python
mae = mean_absolute_error(
    test,
    naive_predictions
)

rmse = np.sqrt(
    mean_squared_error(
        test,
        naive_predictions
    )
)

print("Naive Model MAE:", mae)
print("Naive Model RMSE:", rmse)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    test.index,
    test.values,
    marker='o',
    label='Actual Sales'
)

plt.plot(
    naive_predictions.index,
    naive_predictions.values,
    marker='o',
    label='Naive Predictions'
)

plt.title('Naive Forecast vs Actual Sales - Store 20')
plt.xlabel('Date')
plt.ylabel('Weekly Sales')

plt.legend()
plt.grid(True)
plt.show()
```

```python
window = 4

moving_average_value = train.tail(window).mean()

moving_average_predictions = pd.Series(
    [moving_average_value] * len(test),
    index=test.index
)

print(moving_average_predictions)
```

```python
ma_mae = mean_absolute_error(
    test,
    moving_average_predictions
)

ma_rmse = np.sqrt(
    mean_squared_error(
        test,
        moving_average_predictions
    )
)

print("Moving Average MAE:", ma_mae)
print("Moving Average RMSE:", ma_rmse)
```

```python
model_comparison = pd.DataFrame({
    'Model': ['Naive Forecast', 'Moving Average'],
    'MAE': [mae, ma_mae],
    'RMSE': [rmse, ma_rmse]
})

print(model_comparison)
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
```

```python
holt_model = ExponentialSmoothing(
    train,
    trend='add',
    seasonal=None,
    initialization_method='estimated'
)

holt_fit = holt_model.fit(optimized=True)
```

```python
holt_predictions = holt_fit.forecast(12)

print(holt_predictions)
```

```python
holt_mae = mean_absolute_error(test, holt_predictions)

holt_rmse = np.sqrt(
    mean_squared_error(test, holt_predictions)
)

print("Holt's Method MAE:", holt_mae)
print("Holt's Method RMSE:", holt_rmse)
```

```python
window = 4

history = list(train.values)
moving_average_values = []

for i in range(len(test)):
    prediction = np.mean(history[-window:])
    moving_average_values.append(prediction)

    # Use the prediction for the next forecasting step
    history.append(prediction)

moving_average_predictions = pd.Series(
    moving_average_values,
    index=test.index
)

ma_mae = mean_absolute_error(
    test,
    moving_average_predictions
)

ma_rmse = np.sqrt(
    mean_squared_error(
        test,
        moving_average_predictions
    )
)
```

```python
model_comparison = pd.DataFrame({
    'Model': [
        'Naive Forecast',
        'Moving Average',
        "Holt's Exponential Smoothing"
    ],
    'MAE': [
        mae,
        ma_mae,
        holt_mae
    ],
    'RMSE': [
        rmse,
        ma_rmse,
        holt_rmse
    ]
})

print(model_comparison.sort_values('RMSE'))
```

```python
from statsmodels.tsa.arima.model import ARIMA

arima_model = ARIMA(
    train,
    order=(1, 1, 1)
)

arima_fit = arima_model.fit()

print(arima_fit.summary())
```

```python
arima_predictions = arima_fit.forecast(steps=12)

print(arima_predictions)
```

```python
import pandas as pd
import numpy as np

from sklearn.metrics import mean_absolute_error, mean_squared_error

from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model import ARIMA
```

```python
def generate_forecast(train_data, model_name, horizon=12):

    if model_name == 'Naive':
        return pd.Series(
            [train_data.iloc[-1]] * horizon
        )

    elif model_name == 'Moving Average':

        history = list(train_data.values)
        predictions = []

        for _ in range(horizon):
            prediction = np.mean(history[-4:])
            predictions.append(prediction)
            history.append(prediction)

        return pd.Series(predictions)

    elif model_name == 'Holt':

        model = ExponentialSmoothing(
            train_data,
            trend='add',
            seasonal=None,
            initialization_method='estimated'
        )

        fitted_model = model.fit(optimized=True)

        return fitted_model.forecast(horizon).reset_index(drop=True)

    elif model_name == 'ARIMA':

        model = ARIMA(
            train_data,
            order=(1, 1, 1)
        )

        fitted_model = model.fit()

        return fitted_model.forecast(
            steps=horizon
        ).reset_index(drop=True)

    else:
        raise ValueError("Unknown model")
```

```python
candidate_models = [
    'Naive',
    'Moving Average',
    'Holt',
    'ARIMA'
]

selection_results = []

for store_id in sorted(df['Store'].unique()):

    store_data = (
        df[df['Store'] == store_id]
        .sort_values('Date')
        .set_index('Date')['Weekly_Sales']
    )

    # Split into training, validation and testing
    training = store_data.iloc[:119]
    validation = store_data.iloc[119:131]
    testing = store_data.iloc[131:]

    model_scores = []

    for model_name in candidate_models:

        try:
            predictions = generate_forecast(
                training,
                model_name,
                horizon=12
            )

            predictions.index = validation.index

            model_rmse = np.sqrt(
                mean_squared_error(
                    validation,
                    predictions
                )
            )

            model_scores.append({
                'Model': model_name,
                'RMSE': model_rmse
            })

        except Exception as error:
            print(
                f"Store {store_id}, {model_name}: {error}"
            )

    if not model_scores:
        print(f"No model succeeded for Store {store_id}")
        continue

    best_model = min(
        model_scores,
        key=lambda x: x['RMSE']
    )

    # Evaluate the selected model on the separate test period
    test_training = store_data.iloc[:131]

    test_predictions = generate_forecast(
        test_training,
        best_model['Model'],
        horizon=12
    )

    test_mae = mean_absolute_error(
        testing,
        test_predictions
    )

    test_rmse = np.sqrt(
        mean_squared_error(
            testing,
            test_predictions
        )
    )

    selection_results.append({
        'Store': store_id,
        'Selected_Model': best_model['Model'],
        'Validation_RMSE': best_model['RMSE'],
        'Test_MAE': test_mae,
        'Test_RMSE': test_rmse
    })

model_results = pd.DataFrame(selection_results)

print(model_results)
```
